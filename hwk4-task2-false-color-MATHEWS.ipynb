#### 4. Restoring geospatial information

```python
# a. Use `rio.crs` to print the CRS of this dataset. Is this a geospatial object?
print(landsat.rio.crs)
```

> After running `.rio.crs` on the landsat dataset we got a return value of "none". This means that rioxarray doesn’t see the dataset as geospatial because the CRS information only exists in the file within the `spatial_ref` variable.

```python
# b. You may have noticed in section 3 that the geospatial information for this dataset is stored in the variable `spatial_ref`.

# Print the CRS by accessing the `spatial_ref.crs_wkt` attribute of the dataset.
print(landsat.spatial_ref.crs_wkt)

# Save CRS as a new variable
landsat_crs = landsat.spatial_ref.crs_wkt
```

```python
# c. Recover the geospatial information by using `rio.write_crs()` and the spatial reference information from part b.

# Save the WKT CRS into landsat data
landsat = landsat.rio.write_crs(landsat_crs)

# Print CRS to verify
print(landsat.rio.crs)
```

```python
# a. Without creating any new variables:
# Select the red, green, and blue variables (in that order) of the `xarray.Dataset` holding the Landsat data
# Convert it to an `xarray.DataArray` using the `to_dataarray()` method, and then
# Use `.plot.imshow()` to create an RGB image with the data. There will be a warning; that’s ok.

# Select red, green, blue bands to convert to a DataArray and plot as an RGB image
landsat[['red', 'green', 'blue']].to_dataarray().plot.imshow()
```

```python
# b. Adjust the scale used for plotting the bands to get a true color image. HINT: Check the robust parameter.

# Plot true color image by using robust=True (scale colors and ignore extreme outliers)

# Initialize empty figure
fig, ax = plt.subplots(figsize=(10,5))


# Select red, green, blue bands to convert to a DataArray and plot as an RGB image
landsat[['red', 'green', 'blue']].to_dataarray().plot.imshow(ax=ax, robust=True)

# Customizing title
plt.title('Palisade and Eaton Fire Scars in True Color Imagery (Landsat 8, February 23, 2025)')

# Customizing y-axis
plt.ylabel('y coordinate of projection (meter)')

# Customizing x-axis
plt.xlabel('x coordinate of projection (meter)')

# Display plot
plt.show()
```

```python
# c. To resolve the other warning, identify which bands have `nan` values. 
# HINT: There are many ways of doing so. One option is to use `numpy.isnan()`.

# Made a list to iterate over
bands = ['red', 'green', 'blue', 'nir08', 'swir22']

# Create a for-loop with f-string to print nan values
for band in bands:
    na_val = int(np.isnan(landsat[band]).sum())
    print(f"The {band} has {na_val} NA values.")
```

```python
# d. Use the `.fillna()` method for `xarray.Datasets` to replace any nan values in the Landsat data with zero.

# Replace landsat NA values with 0
landsat = landsat.fillna(value = 0)
```

```python
# e. Create a new true color image that gets plotted without warnings.

# Plot true color image with no warnings

# Initialize empty figure
fig, ax = plt.subplots(figsize=(10,5))

# Select red, green, blue bands to convert to a DataArray and plot as an RGB image
landsat[['red', 'green', 'blue']].to_dataarray().plot.imshow(ax=ax, robust=True)

# Customizing title
plt.title('Palisade and Eaton Fire Scars in True Color Imagery (Landsat 8, February 23, 2025)')

# Customizing y-axis
plt.ylabel('y coordinate of projection (meter)')

# Customizing x-axis
plt.xlabel('x coordinate of projection (meter)')

# Display plot
plt.show()
```

#### 7. Map

a. Create a map showing the shortwave infrared/near-infrared/red false color image together with both fire *perimeters*. Customize it appropriately including, at least, an informative title and legend. You may also want to include text on the map to identify which fire is which.

b. Write a figure description for the map including a brief explanation of how false color imagery is being used.

```python
# Data preparation

# Change Eaton crs to landsat crs
eaton = eaton.to_crs(landsat.rio.crs)

# Change palisades crs to landsat crs
palisades = palisades.to_crs(landsat.rio.crs)

# Check crss' match
assert eaton.crs == landsat.rio.crs
assert eaton.crs == landsat.rio.crs
```

```python
# Join datasets
fires = pd.concat([eaton, palisades], ignore_index=True)

# Find max point for each fire
eaton_max_coord = eaton.geometry.get_coordinates().max()
palisades_max_coord = palisades.geometry.get_coordinates().max()

# Save max point into a variable
max_x = [eaton_max_coord.iloc[0], palisades_max_coord.iloc[0]]
max_y = [eaton_max_coord.iloc[1], palisades_max_coord.iloc[1]]
```

```python
# a. Create a map showing the shortwave infrared/near-infrared/red false color image together with both fire *perimeters*.
# Customize it appropriately including, at least, an informative title and legend. 
# You may also want to include text on the map to identify which fire is which.

# Initialize empty figure
fig, ax = plt.subplots(figsize=(10,5))

# Select swir22, nir08, red bands to convert to a DataArray and plot as an RGB image
landsat[['swir22', 'nir08', 'red']].to_dataarray().plot.imshow(ax=ax, robust=True)

# Plot Eaton Fire boundary
eaton.boundary.plot(ax=ax, color='red', label='Eaton Fire', linewidth=0.75)

# Plot Eaton Fire boundary
palisades.boundary.plot(ax=ax, color='orange', label='Palisades Fire', linewidth=0.75)

# Customize title
plt.title('Eaton and Palisades Fire Scars in False Color (Landsat 8, February 23, 2025)')

# Customize y-axis
plt.ylabel('y coordinate of projection (meter)')

# Customize x-axis
plt.xlabel('x coordinate of projection (meter)')

# Show legend
plt.legend()

# Add fire names as text annotations
for x, y, name in zip(max_x, max_y, fires['FIRE_NAME']):
    ax.text(x, y - 6000, name, fontsize=10, color = 'white', fontweight = 'bold', ha='left')

# Display plot
plt.show()
```

> The map shows a false color image of the Los Angeles area captured by Landsat 8 on February 23, 2025, overlaid with the perimeters of the Palisades Fire (orange) and Eaton Fire (red). The image assigns the shortwave infrared (`swir22`) to the red channel, near-infrared (`nir08`) to the green channel, and red band to the blue channel. Healthy vegetation strongly reflects near-infrared, so it appears bright green, while burned areas reflect shortwave infrared more than near-infrared, causing them to appear reddish-brown. This contrast makes the fire scars stand out against the surrounding unaffected areas, and the scars align closely with the mapped fire perimeters.
