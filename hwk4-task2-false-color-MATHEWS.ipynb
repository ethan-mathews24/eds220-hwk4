> After running `.rio.crs` on the landsat dataset we got a return value of "none". This means that rioxarray doesn’t see the dataset as geospatial because the CRS information only exists in the file within the `spatial_ref` variable.

```python
# b. You may have noticed in section 3 that the geospatial information for this dataset is stored in the variable `spatial_ref`.

# Print the CRS by accessing the `spatial_ref.crs_wkt` attribute of the dataset.
print(landsat.spatial_ref.crs_wkt)

# Save CRS as a new variable
landsat_crs = landsat.spatial_ref.crs_wkt
```

```python
# c. Recover the geospatial information by using `rio.write_crs()` and the spatial reference information from part b.

# Save the WKT into landsat data
landsat = landsat.rio.write_crs(landsat_crs)

# Print CRS to verify
print(landsat.rio.crs)
```
